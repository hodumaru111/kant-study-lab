# 데이터의 첫인상 읽기

## CSV 파일과 pandas DataFrame의 차이

* csv
csv는 하나의 파일
ex) customer_id,name,gender,age,city,signup_date
    1,김철수,M,34,서울,2024-02-10
    2,이영희,F,29,부산,2023-11-05
* padas
pandas는 데이터를 분류하고 문제 있는 데이터를 골라냄
* DataFrame
pandas의 read_csv()는 csv 파일을 읽어 DataFrame으로 만듦
즉 요리를 하기 위해 csv(식재료)을 도마 위에 올려둔 상태 

| Index (인덱스) | customer_id (열) | name (열) | gender (열) | age (열) | city (열) | signup_date (열) |
| :---: | :---: | :---: | :---: | :---: | :---: | :---: |
| **0** (행 1) | 1001 | 김철수 | M | 29 | 서울 | 2023-01-15 |
| **1** (행 2) | 1002 | 이영희 | F | 34 | 부산 | 2023-02-20 |
| **2** (행 3) | 1003 | 박민수 | M | 41 | 대구 | 2023-03-05 |
| **3** (행 4) | 1004 | 최지지원 | F | 25 | 인천 | 2023-04-12 |

## 행, 열, 값, 인덱스와 업무용 ID를 구분

* 행
한 행은 고객 한 명을 의미

* 열
고객의 속성을 의미

* 값
특정 행과 열이 만나는 실제 데이터
---예를 들어 첫 번째 고객의 city 값은 서울입니다.---

* 인덱스 
데이터를 뻐르게 조회하고 정렬하기 위해 행에 붙여둔 '가변적 주소'

* 업무용 ID
CSV 파일, 데이터베이스..등 '진짜 데이터' 고유성을 유지

*인덱스와 업무용 ID의 차이
인덱스는 CSV파일을 읽어와 DataFrame을 만든 순간에만 존재하게 되는 '임시 번호'
