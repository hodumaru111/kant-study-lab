## 행, 열, 값, 인덱스와 업무용 ID를 구분

* 행 (Row)

.
* 열 (Column)

  고객의 속성을 의미

.
* 값

.

  CSV 파일, 데이터베이스..등 '진짜 데이터' 고유성을 유지

# shape, head(), tail(), columns, info(), dtypes의 역할 구분

# 첫 점검은 넓게 보고 점점 좁혀 감



파일 존재

.

1단계: 재료 입고 확인 (파일 존재 → DataFrame 로드)
* 어떤 작업인가: 데이터 파일(CSV 등)이 진짜 지정한 폴더에 있는지 확인하고, 파이썬이 읽을 수 있는 표 형태로 메모리에 올리는 과정임

* 왜 하나: 파일 이름에 오타가 있거나 경로가 틀리면 다음 작업으로 아예 넘어가지도 못하고 에러가 나거든. 문이 열려야 집 안을 둘러볼 수 있는 것과 같음

.

2단계: 겉모습 스캔 (shape → head / tail → columns)
* 어떤 작업인가: 표의 전체적인 방 크기와 형태를 훑어보는 단계

* shape: 이 표가 총 몇 행(세로줄) 몇 열(가로줄)짜리 대형 표인지 체급을 확인

* head / tail: 맨 위 5줄, 맨 아래 5줄을 미리 짚어보면서 한글이 깨지진 않았는지, 파일 맨 밑에 쓸데없는 합계 줄이 껴 있진 않은지 맛보기로 봄

* columns: 열 이름에 띄어쓰기나 이상한 특수문자가 들어가서 나중에 코딩할 때 방해되지 않는지 점검

.

3단계: 기본 체력 진단 (info/dtypes → 결측치 → 전체 중복 → 주요 ID)
* 어떤 작업인가: 데이터가 오염되었거나 구멍이 뚫려 있는지 확인하는 필수 정밀 검사

* info / dtypes: 숫자로 계산해야 할 데이터(예: 가격, 나이)가 글자(문자열)로 잘못 읽히지 않았는지 확인

* 결측치: 비어 있는 칸(NaN)이 어디에 얼마나 뚫려 있는지 파악해. 빈칸을 그냥 두면 나중에 평균이나 합계를 낼 때 오류 남

* 전체 중복: 시스템 오류로 완전히 똑같은 행이 두 번 복사되어 들어간 게 없는지 찾아냄

* 주요 ID: 주민번호나 주문번호처럼 '절대 중복되면 안 되고 빈칸도 없어야 하는 핵심 열'이 멀쩡한지 검증해. 이 고유 키가 꼬이면 나중에 데이터 전체가 엉망이 됨

.


4단계: 디테일 및 관계 검사 (숫자·범주·날짜 → 파일 간 관계)
* 어떤 작업인가: 데이터 안의 진짜 내용물이 현실적인지, 다른 표들과 짝이 잘 맞는지 보는 마지막 단계야.

* 숫자·범주·날짜:

* 숫자: 나이가 -10살이거나 제품 가격이 0원인 비상식적인 값이 있는지 확인해.

* 범주: 성별 열에 남, 남자, M처럼 제멋대로 섞여 적힌 게 없는지 묶어봐.

* 날짜: 2099년이나 1900년 같은 말도 안 되는 날짜 데이터나 글자 형식으로 굳어버린 날짜를 찾아내.

* 파일 간 관계: '고객 표'와 '주문 표'를 연결할 때, 고객 목록에는 없는데 주문 표에만 뜬금없이 존재하는 '유령 고객의 주문'이 없는지 참조 관계를 조율해.
→ 파일 간 관계

.

* 데이터를 본격적으로 다루기 전 "이 데이터는 믿고 분석해도 된다"는 최소한의 안전장치를 확보하는 과정

# 전체 행 중복과 업무용 ID 중복을 구분

## 전체 행 중복

전체 행이 완전히 같은지 확인

 

customers.duplicated().sum()

 

전체 중복은 데이터 적재 과정에서 같은 행이 두 번 들어간 경우일 수 있음

 

하지만 삭제 전 원인을 확인해야 함

## 주요 ID 중복

customers["customer_id"].duplicated().sum()

 

고객 한 명당 한 행이라는 규칙이라면 customer_id 중복은 문제일 가능성이 높습니다.

 

그러나 다음 컬럼의 중복은 정상일 수 있습니다.

 

orders.customer_id

order_items.order_id

order_items.product_id

 

* 한 고객은 여러 주문을 할 수 있고, 한 주문에는 여러 상품이 들어갈 수 있기 때문입니다.


## 결측과 중복을 함께 확인


id_missing = customers["customer_id"].isna().sum()
id_duplicates = customers["customer_id"].duplicated().sum()
print("ID 결측:", id_missing)
print("ID 중복:", id_duplicates)

* isna = 빈칸이면 1 채워져있으면 0
* sum = isna가 찾은 거에 숫자를 부여해 보여줌
* duplicates = 파일 내 중복 찾아줌
